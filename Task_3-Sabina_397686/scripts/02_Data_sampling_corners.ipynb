# IMPORT LIBRARIES

```python
## To import necessary libraries
import pandas as pd
from pathlib import Path
from common import *
```

# LOAD PROCESSED DATA

```python
## To load the processed data from CSV file
df_long = pd.read_csv("../data/processed data/fifaworldcup2026_team_match_corners_prepared.csv")
df_long.head()
```

# DATA SAMPLING

```python
## To seperate the corners data for group stage and knockout stage
group_corners = df_long[df_long["stage"] == "Group"]["corners"]
knockout_corners = df_long[df_long["stage"] == "Knockout"]["corners"]

print("Group n:", len(group_corners))
print("Knockout n:", len(knockout_corners))
```

```python
# To set the sample size for group and knockout stages
sample_size_group = 50      # choose a sample size, must be ≥30
sample_size_knockout = 50   # can't exceed available n=62 for knockout

# To extract simple random samples as clean numerical arrays
group_sample = group_corners.sample(n=sample_size_group, random_state=42)
knockout_sample = knockout_corners.sample(n=sample_size_knockout, random_state=42)

 # To verify extraction properties
print("Group sample n:", len(group_sample))
print("Knockout sample n:", len(knockout_sample))
```

# SAVE SAMPLE DATA

```python
## To check if the sample files already exist and avoid overwriting them and save the samples to CSV files if they don't exist
group_path = Path("../data/sample data/sample_group_corners.csv")
knockout_path = Path("../data/sample data/sample_knockout_corners.csv")

if not group_path.exists() and not knockout_path.exists():
    group_sample.to_csv(group_path, index=False, header=["corners"])
    knockout_sample.to_csv(knockout_path, index=False, header=["corners"])
    print("Sample files created.")
else:
    print("Sample files already exist — skipping regeneration.")
```

# LOG OUTPUT SUMMARY INTO TEXT FILE

```python
# To save the output of the data sampling process to a text file
output_file = OUTPUT_FOLDER / "02_data_sampling_output.txt"

group_corners, knockout_corners, group_sample, knockout_sample = generate_samples()

output = f"""

DATA SAMPLING

Population sizes:
Group n: {len(group_corners)}
Knockout n: {len(knockout_corners)}

Sample sizes chosen:
Group sample size: {SAMPLE_SIZE_GROUP}
Knockout sample size: {SAMPLE_SIZE_KNOCKOUT}

Sample extraction verified:
Group sample n: {len(group_sample)}
Knockout sample n: {len(knockout_sample)}

Sampling completed successfully.
"""

output_file.write_text(output, encoding="utf-8")
print(output)
print("\nOutput automatically saved to:")
print(output_file)
```
