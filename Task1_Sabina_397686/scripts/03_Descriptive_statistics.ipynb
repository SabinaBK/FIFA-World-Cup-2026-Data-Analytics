# IMPORT LIBRARIES

```python
## To import necessary libraries
import pandas as pd
import statistics as st
import numpy as np
from common import *
```

# LOAD SAMPLE DATA 

```python
## To load the sampled data from CSV files
group_corners = pd.read_csv("../data/sample data/sample_group_corners.csv")["corners"]
knockout_corners = pd.read_csv("../data/sample data/sample_knockout_corners.csv")["corners"]

print("Group sample n:", len(group_corners))
print("Knockout sample n:", len(knockout_corners))
```

# DESCRIPTIVE STATISTICS

```python
# DESCRIPTIVE STATISTICS — GROUP STAGE
sample = list(group_corners)

## mean
x_bar = st.mean(sample)
print("Mean:", x_bar)

## median
median = st.median(sample)
print("Median: %.2f" % median)

## mode
mode = st.mode(sample)
print("Mode value: %d" % mode)

## range
the_range = np.max(sample) - np.min(sample)
print("Range: %d" % the_range)

## sample and population variance
sample_arr = np.array(sample)
s_square = sample_arr.var(ddof=1)
sigma_square = sample_arr.var()
print("Sample variance: %.2f. Population variance: %.2f." % (s_square, sigma_square))

## sample and population standard deviation
s = sample_arr.std(ddof=1)
sigma = sample_arr.std()
print("Sample std. dev.: %.2f. Population std. dev.: %.2f." % (s, sigma))

## iqr and percentiles
pct25 = np.percentile(sample, 25)
pct75 = np.percentile(sample, 75)
iqr = pct75 - pct25
print("IQR: %d. 25th percentile: %d. 75th percentile: %d" % (iqr, pct25, pct75))
```

```python
# DESCRIPTIVE STATISTICS — KNOCKOUT STAGE
sample = list(knockout_corners)

## mean
x_bar = st.mean(sample)
print("Mean:", x_bar)

## median
median = st.median(sample)
print("Median: %.2f" % median)

## mode
mode = st.mode(sample)
print("Mode value: %d" % mode)

## range
the_range = np.max(sample) - np.min(sample)
print("Range: %d" % the_range)

## sample and population variance
sample_arr = np.array(sample)
s_square = sample_arr.var(ddof=1)
sigma_square = sample_arr.var()
print("Sample variance: %.2f. Population variance: %.2f." % (s_square, sigma_square))

## sample and population standard deviation
s = sample_arr.std(ddof=1)
sigma = sample_arr.std()
print("Sample std. dev.: %.2f. Population std. dev.: %.2f." % (s, sigma))

## iqr and percentiles
pct25 = np.percentile(sample, 25)
pct75 = np.percentile(sample, 75)
iqr = pct75 - pct25
print("IQR: %d. 25th percentile: %d. 75th percentile: %d" % (iqr, pct25, pct75))
```
