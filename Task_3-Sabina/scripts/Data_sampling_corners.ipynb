# IMPORT LIBRARIES

```python
## To import necessary libraries
import pandas as pd
```

# LOAD PROCESSED DATA

```python
## To load the processed data from CSV file
df_long = pd.read_csv("../data/processed data/fifaworldcup2026_team_match_corners_prepared.csv")
df_long.head()
```

# DATA SAMPLING

```python
## To seperate the corners data for group stage and knockout stage
group_corners = df_long[df_long["stage"] == "Group"]["corners"]
knockout_corners = df_long[df_long["stage"] == "Knockout"]["corners"]

print("Group n:", len(group_corners))
print("Knockout n:", len(knockout_corners))
```

```python
## To set the sample size for group and knockout stages
sample_size_group = 50      # choose a sample size, must be ≥30
sample_size_knockout = 50   # can't exceed available n=62 for knockout

group_sample = group_corners.sample(n=sample_size_group, random_state=42)
knockout_sample = knockout_corners.sample(n=sample_size_knockout, random_state=42)

print("Group sample n:", len(group_sample))
print("Knockout sample n:", len(knockout_sample))
```

```python
## To save the sampled data to CSV files
group_sample.to_csv("sample_group_corners.csv", index=False, header=["corners"])
knockout_sample.to_csv("sample_knockout_corners.csv", index=False, header=["corners"])
```
