# DATA WRANGLING

```python
## To reshape to long format (one row per team per match)
df_team1 = df_matches[["date", "stage", "round", "team1", "corners_t1"]].rename(
    columns={"team1": "team", "corners_t1": "corners"}
)
df_team2 = df_matches[["date", "stage", "round", "team2", "corners_t2"]].rename(
    columns={"team2": "team", "corners_t2": "corners"}
)

df_long = pd.concat([df_team1, df_team2], ignore_index=True)
print(df_long.shape)
print(df_long.head())
```

```python
### To check for any missing values in the DataFrame
print(df_long.isnull().sum())
```

```python
## To verify total number of matches in each stage and round after the transformation
print(df_long.shape)
print(df_long["stage"].value_counts())
```
