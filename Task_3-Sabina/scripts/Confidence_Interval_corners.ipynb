# IMPORT LIBRARIES

```python
## To import necessary libraries
import pandas as pd
import math
import scipy.stats as st
```

# LOAD SAMPLE DATA

```python
## To load the sampled data from CSV files
group_corners = pd.read_csv("../data/sample data/sample_group_corners.csv")["corners"]
knockout_corners = pd.read_csv("../data/sample data/sample_knockout_corners.csv")["corners"]

print("Group sample n:", len(group_corners))
print("Knockout sample n:", len(knockout_corners))
```

# CONFIDENCE INTERVAL

```python
# CONFIDENCE INTERVAL — GROUP STAGE
## sample mean, standard deviation, and sample size
x_bar = group_corners.mean()
s = group_corners.std(ddof=1)
n = len(group_corners)
print("Mean: %.2f. Standard deviation: %.2f. Size: %d." % (x_bar, s, n))

## z-score (assuming 95% Confidence Level)
z_score = st.norm.ppf(q=0.975)
print("Z-statistic: %.2f" % z_score)

## compute standard error
std_err = s / math.sqrt(n)
print("Standard error: %.2f" % std_err)

## compute the margin of error
mrg_err = z_score * std_err
print("Margin of error: %.2f" % mrg_err)

## get the lower and upper bound of confidence interval
ci_low = x_bar - mrg_err
ci_upp = x_bar + mrg_err
print("Confidence Interval of the mean: %.2f to %.2f" % (ci_low, ci_upp))
```

```python
# CONFIDENCE INTERVAL — KNOCKOUT STAGE
## sample mean, standard deviation, and sample size
x_bar = knockout_corners.mean()
s = knockout_corners.std(ddof=1)
n = len(knockout_corners)
print("Mean: %.2f. Standard deviation: %.2f. Size: %d." % (x_bar, s, n))

## z-score (assuming 95% Confidence Level)
z_score = st.norm.ppf(q=0.975)
print("Z-statistic: %.2f" % z_score)

## compute standard error
std_err = s / math.sqrt(n)
print("Standard error: %.2f" % std_err)

## compute the margin of error
mrg_err = z_score * std_err
print("Margin of error: %.2f" % mrg_err)

## get the lower and upper bound of confidence interval
ci_low = x_bar - mrg_err
ci_upp = x_bar + mrg_err
print("Confidence Interval of the mean: %.2f to %.2f" % (ci_low, ci_upp))
```
